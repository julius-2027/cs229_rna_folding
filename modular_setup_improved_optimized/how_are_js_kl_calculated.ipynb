```python
from scipy.spatial.distance import jensenshannon
from scipy.stats import wasserstein_distance
import numpy as np
import matplotlib.pyplot as plt

# bin_edges = np.load('../modular_setup_improved/bin_edges.npy')
# bin_centers = 0.5*(bin_edges[1:] + bin_edges[:-1])
```

```python
all_targets = np.array([[1.0, 0.0, 0.0],[0.3,0.3,0.4],[0.3,0.5,0.2]])
all_preds = np.array([[0.0, 1.0, 0.0],[0.8,0.05,0.15],[0.3,0.5,0.2]])

eps = 1e-8
safe_preds = np.clip(all_preds, eps, None)
safe_preds = safe_preds / safe_preds.sum(axis=-1, keepdims=True)

js_scores = np.array(
    [jensenshannon(all_targets[i], all_preds[i]) for i in range(len(all_targets))]
)

m = 0.5*(all_targets + all_preds)
safe_m = np.clip(m, eps, None)
safe_m = safe_m / safe_m.sum(axis=-1, keepdims=True)
with np.errstate(divide="ignore", invalid="ignore"):
    js_terms = (np.where(all_targets > 0, all_targets * (np.log(all_targets) - np.log(safe_m)), 0.0)
                + np.where(all_preds > 0, all_preds * (np.log(all_preds) - np.log(safe_m)), 0.0))
js_scores_byhand = np.sqrt(0.5*js_terms.sum(axis=-1))

with np.errstate(divide="ignore", invalid="ignore"):
    kl_terms = np.where(
        all_targets > 0, all_targets * (np.log(all_targets) - np.log(safe_preds)), 0.0
    )

kl_scores = kl_terms.sum(axis=-1)

wasserstein_scores = np.array(
    [wasserstein_distance([-10,0,10], [-10,0,10], all_targets[i], all_preds[i]) for i in range(len(all_targets))]
)

wasserstein_scores_byhand = np.array(
    [10*np.sum(np.abs(np.cumsum(all_targets[i]) - np.cumsum(all_preds[i]))) for i in range(len(all_targets))]
)
```

```python
print(js_scores)
print(js_scores_byhand)
print(kl_scores)
print(wasserstein_scores)
print(wasserstein_scores_byhand)
```
